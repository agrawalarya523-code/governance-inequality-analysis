```r
setwd("../data/processed")
long_form <- read.csv("long_form.csv")

# Model objects get saved here so plot-generation.ipynb can plot the exact fitted models
# (and their standard errors) instead of silently re-fitting its own copy
dir.create("models", showWarnings = FALSE)
```

## Bivariate relationships with GINI

Each predictor below (education spending, corruption, business-regulation quality) is tested
**independently** against `GINI` via a Pearson correlation and a simple linear regression.
These are univariate estimates — the predictors aren't tested jointly, so any correlation
between them (e.g. more corrupt countries may also score lower on regulation quality) isn't
controlled for here. Read each coefficient as "the raw association with GINI," not as an
effect isolated from the other two.

```r
# --- Education spend ---
cor_education <- cor.test(long_form$GINI, long_form$Education_Spend, method = "pearson")
cor_education
```

```r
model_education <- lm(GINI ~ Education_Spend, data = long_form)
saveRDS(model_education, "models/model_education.rds")
summary(model_education)
```

```r
# --- Corruption ---
cor_corruption <- cor.test(long_form$GINI, long_form$Corruption, method = "pearson")
cor_corruption
```

```r
model_corruption <- lm(GINI ~ Corruption, data = long_form)
saveRDS(model_corruption, "models/model_corruption.rds")
summary(model_corruption)
```

```r
# --- Regulation ---
cor_regulation <- cor.test(long_form$GINI, long_form$Regulation, method = "pearson")
cor_regulation
```

```r
model_regulation <- lm(GINI ~ Regulation, data = long_form)
saveRDS(model_regulation, "models/model_regulation.rds")
summary(model_regulation)
```
